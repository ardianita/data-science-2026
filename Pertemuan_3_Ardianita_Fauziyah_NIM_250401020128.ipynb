```python
import pandas as pd
import numpy as np
from scipy.stats.mstats import winsorize

# Memuat Dataset
df = pd.read_csv("https://drive.google.com/uc?id=1LfQWProB0VjWN5q8bKuRIgn-stULfIRo")
df
```

```python
# Eksplorasi Dataset

print('Shape awal:')
print(df.shape);

print(f'\nTipe data:');
df.info();

print(f'\nStatistik deskripsi:')
print(df.describe());

print(f'\nMissing value')
print(df.isnull().sum());
```

```python
# Hapus duplikasi
df.drop_duplicates(inplace=True);
print(f"Total Data Duplikat : {df.duplicated().sum()}")
print(f"\nShape Akhir")
print(f"Baris: {df.shape[0]}");
print(f"Kolom: {df.shape[1]}");
```

```python
# Normalisasi String

# Normalisasi nama kota dengan capital di huruf awal
df['kota']    = df['kota'].str.strip().str.title()

# Normalisasi kondisi menjadi huruf kecil
df['kondisi'] = df['kondisi'].str.strip().str.lower()

print("Dataset setelah normalisasi")
df
```

```python
# Imputasi missing values

# Imputasi median value luas
df['luas_m2'] = df['luas_m2'].fillna(df['luas_m2'].median())

# Imputasi median value harga_jual
df['harga_juta'] = df['harga_juta'].fillna(df['harga_juta'].median())

# Imputasi modus value kamar
df['kamar'] = df['kamar'].fillna(df['kamar'].mode()[0])

# Dataset setelah imputasi
print('Dataset Setelah Imputasi')
df
```

```python
# Menangani outlier dengan IQR
for col in ['harga_juta', 'luas_m2']:
    Q1, Q3 = df[col].quantile([0.25, 0.75])

    # IQR
    IQR = Q3 - Q1

    # Batas bawah & atas
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    # Outlier
    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"\nKolom: {col}")
    print(f"Batas bawah: {lower:.0f}")
    print(f"Batas atas : {upper:.0f}")
    print(f"Jumlah outlier: {len(outliers)} baris")
```

```python
# Validasi Akhir

print(f'Missing value')
print(df.isnull().sum())

duplikat = df.duplicated().sum()
print(f'\nTotal Duplikasi Data : {duplikat}')
```

```python
df.to_csv('housing_clean.csv', index=False)
print('Dataset bersih tersimpan!')
```
